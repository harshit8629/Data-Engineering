```python
# I4
marks = [35, 78, 42, 90, 28, 55, 67, 39]

# Create a second list with only qualifying marks (>= 40)
qualifying_marks = [mark for mark in marks if mark >= 40]

# Display the qualifying marks and their count
print("Qualifying marks:", qualifying_marks)
print("Number of students who qualified:", len(qualifying_marks))
```

```python
# C5
daily_sales = [800,1000,1500,1300,1100,980,2000,1800,1700,6000,
              4500,6589,458,4926,9546,3654,8246,8165,4682,6245,
              4548,6526,3265,2326,5133,6263,2323,2632,2323,6132]

total=sum(daily_sales)
print("Total Sales: ",total)

# Average
average_sales=sum(daily_sales)/len(daily_sales)
print("Average_daily_sales: ",average_sales)

# Highest sales
highest = max(daily_sales)
print("Highest Sales: ",highest)

# the number of days when sales exceeded the average 
exceeded_average = [sale for sale in daily_sales if sale>average_sales]
print('The Number of days when sales exceeded the average: ', len(exceeded_average))
```
